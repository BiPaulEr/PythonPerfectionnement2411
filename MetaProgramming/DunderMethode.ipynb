```python
class Vector:
     def __init__(self, x, y, z):
         self.x = x
         self.y = y
         self.z = z
     def __repr__(self):
         return f"<Vector: x={self.x}, y={self.y}, z={self.z}>"
     def __eq__(self, other):
         return self.x == other.x and self.y == other.y and self.z == other.z

vec1 = Vector(1, 3, 6)
vec2 = Vector(1, 3, 6)
print(vec1)
print(vec1 == vec2)
```

```python
from dataclasses import dataclass
@dataclass
class VectorDataClass:
    x: int
    y: int 
    z: int
    
    def __le__(self, other):
        return self.x < other.x 
    def __lt__(self, other):
        return self.x <= other.x 

vec1 = VectorDataClass(1, 3, 6)
vec2 = VectorDataClass(1, 3, 6)
print(vec1 <= vec2)
print(vec1 < vec2)
print(vec1 == vec2)
```

```python
from dataclasses import dataclass
@dataclass(order=True)
class VectorDataClass:
    x: int
    y: int 
    z: int

vec1 = VectorDataClass(1, 3, 6)
vec2 = VectorDataClass(1, 3, 6)
print(vec1 <= vec2)
print(vec1 < vec2)
print(vec1 > vec2)
print(vec1 >= vec2)
print(vec1 == vec2)
```

```python
class MyDict(dict):
    def __setitem__(self, key, value):
        print(f"Setting {key} = {value}")
        super().__setitem__(key, value)
        
d = MyDict(a=1)
d['b'] = 2
print(d)
d.update({'c': 3})
```

```python
from collections import UserDict
class MyDict(UserDict):
    def __setitem__(self, key, value):
        print(f"Setting {key} = {value}")
        super().__setitem__(key, value)
        
d = MyDict(a=1)
d['b'] = 2
print(d)
d.update({'c': 3})
```

```python
class A:
    def __init__(self):
        print("A", end=" ")
        super().__init__()

class B:
    def __init__(self):
        print("B", end=" ")
        super().__init__()
        
class C(A, B):
    def __init__(self):
        print("C", end=" ")
        super().__init__()
       
c = C()
```

```python
class A:
    def __init__(self, nombre_de_A, *args):
        self.nombre_de_A = nombre_de_A
        print("A" * nombre_de_A, end=" ")
        super().__init__(*args)

class B:
    def __init__(self, nombre_de_B):
        self.nombre_de_B = nombre_de_B
        print("B" * nombre_de_B, end=" ")
        super().__init__()
        
class C(A, B):
    def __init__(self, nombre_de_A, nombre_de_B):
        print("C", end=" ")
        super().__init__(nombre_de_A, nombre_de_B)
       
c = C(4, 10)
```

```python
class Circle:
    def __init__(self, radius):
            self.__radius = radius

    @property
    def radius(self):
            """Obtenir le rayon du cercle."""
            return self.__radius

    @radius.setter
    def radius(self, value):
            if value <= 0:
                    raise ValueError("Le rayon doit être positif.")
            self.__radius = value
        
    @property
    def diameter(self):
            return self.__radius * 2
circle = Circle(12)
print(circle.radius)

circle.radius = 90
print(circle.radius)
circle.diameter
```
