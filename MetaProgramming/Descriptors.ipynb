```python
class RevealAccess:
    def __init__(self, nom, val):
        self.val = val
        self.nom = nom
    def __get__(self, obj, objtype):
        print(f"Rettriving element {self.nom}")
        print(obj, objtype)
        return self.val
    def __set__(self, obj, val):
        print(f"Setting element {self.nom} by {val}", obj)
        self.val = val

class MyClass:
    y = RevealAccess("y_datadescriptors", 6)

myclass = MyClass()
print(myclass)
myclass.y = 12
myclass.__dict__
print(myclass.y)
```

```python
class IntegerPositif:
    def __init__(self, name):
        self.name = name
        
    def __get__(self, obj, objtype):
        return obj.__dict__.get(self.name, 0)
        
    def __set__(self, obj, val):
        if val > 0 and val < 130:
            obj.__dict__[self.name] = val

class Person:
    age = IntegerPositif("age_protected")

person = Person()
person.age = 6
print(person.age)
person.age = -2
print(person.age)
```

```python
class LazyProperty:
    def __init__(self, func):
        self.func = func
        self.val = None
        
    def __get__(self, obj, objtype):
        if self.val is None:
            print("Calculate")
            self.val = self.func(obj)
        return self.val
        
    def __set__(self, obj, val):
        pass

class Circle:
    radius = 9
    area = LazyProperty(lambda self: self.radius ** 2 * 3.14)

circle = Circle()
```

```python
circle.area
```

```python
class ReadOnly:
    def __init__(self, val):
        self.val = val
        
    def __get__(self, obj, objtype):
        return self.val
        
    def __set__(self, obj, val):
        raise AttributeError("NON")

class Circle:
    radius = ReadOnly(9)

circle = Circle()
print(circle.radius)
circle.radius = 122
```

```python
circle.radius
```

```python
class AttributDynamique:
    def __init__(self, func):
        self.func = func
        
    def __get__(self, obj, objtype):
        return self.func(obj)
        
    def __set__(self, obj, val):
        pass

class Circle:
    radius = 9
    area = AttributDynamique(lambda self: self.radius ** 2 * 3.14)

circle = Circle()
print(circle.area)
circle.radius = 10
print(circle.area)
```
