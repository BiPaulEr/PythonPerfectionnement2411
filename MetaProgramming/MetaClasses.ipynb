```python
class Metaclass(type):
    @classmethod
    def __prepare__(mcs, name, bases, **kwargs):
        print("__prepare__", mcs, name, bases)
        namespace = super().__prepare__(name, bases, **kwargs)
        return namespace


    def __new__(mcs, name, bases, namespace):
        print("__new__", mcs, name, bases)#, namespace)
        return super().__new__(mcs, name, bases, namespace)
   
 
    def __init__(cls, name, bases, namespace, **kwargs):
        print("__init__", cls, name, bases)#, namespace)
        super().__init__(name, bases, namespace)

    def __call__(cls, *args, **kwargs):
        print("__call__", cls)
        return super().__call__(*args, **kwargs)


class InstanceCountingClass(metaclass=Metaclass):
    def __new__(cls, *args, **kwargs):
        print("C__new__()",  cls)
        instance = super().__new__(cls)
        return instance

    def __init__(self, attribute):
        print('C__init__()', self, attribute)
        self.attribute = attribute

instance1 = InstanceCountingClass("oui")
```

```python
class PluginLogin(type):
    list_plugins = []
    @classmethod
    def __prepare__(mcs, name, bases, **kwargs):
        namespace = super().__prepare__(name, bases, **kwargs)
        return namespace

    def __new__(mcs, name, bases, namespace):
        cls = super().__new__(mcs, name, bases, namespace)
        mcs.list_plugins.append(cls)
        return cls
 
    def __init__(cls, name, bases, namespace, **kwargs):
        super().__init__(name, bases, namespace)

    def __call__(cls, *args, **kwargs):
        return super().__call__(*args, **kwargs)


class Plugin1(metaclass=PluginLogin):
    pass
class Plugin2(metaclass=PluginLogin):
    pass
class Plugin3(metaclass=PluginLogin):
    pass
class Plugin4(metaclass=PluginLogin):
    pass   
    
def can_login():
    for element in PluginLogin.list_plugins:
        print(element())
can_login()
```
