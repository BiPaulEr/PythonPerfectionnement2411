```python
import time
def info(thread):
    print("name", thread.name)
    print("ident", thread.ident)
    print("alive", thread.is_alive())
    
def simple_function():
    time.sleep(5)
    print("hello")

t = Thread(target=simple_function)
info(t)
t.start()
info(t)
```

```python
import string
def write_number():
    with open("compte.txt", "a") as file:
        for i in range(100, 1000):
            file.write(str(i)+" ")
            file.flush()
            
def write_letter():
    liste = list(string.ascii_lowercase)[:5]
    with open("compte.txt", "a") as file:
        for i in liste:
            file.write(str(i)+" ")
            file.flush()
            
liste_thread_num = []
liste_thread_let = []
for i in range(0, 10):
    t = Thread(target=write_number)
    liste_thread_num.append(t)
    t.start()
    
for i in range(0, 3):
    t = Thread(target=write_letter)
    liste_thread_let.append(t)
    
for t in liste_thread_num:
    t.join()
    
for t in liste_thread_let:
    t.start()
    

print("end")
```
