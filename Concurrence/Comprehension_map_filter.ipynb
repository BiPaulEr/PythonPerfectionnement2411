```python
liste = [1, 2, 3, 4, 5]
for i in range(0, len(liste)):
    liste[i] = liste[i]**2
print(liste)
```

```python
liste_modifie = map(lambda x : x**2, liste)
print(liste_modifie)
```

```python
liste.append(0)
list(liste_modifie)
```

```python
list(liste_modifie)
```

```python
liste = [1, 2, 3, 4, 5]
filtre1 = filter(lambda x : x%2, liste)
filtre2 = filter(lambda x : False, liste)
print(list(filtre1))
print(list(filtre2))
```

```python
liste = [1, 2, 3, 4, 5]
filtre1 = map(lambda x : x**2, filter(lambda x : x%2, liste))
print(list(filtre1))
```

```python
liste = [1, 2, 3, 4, 5]

filtre1 = [x**2 for x in liste if x%2]
print(filtre1)
```

```python
liste = [1, 2, 3, 4, 5]

filtre1 = (x**2 for x in liste if x%2)
print(filtre1)
```
