```python
class ContextManager():
    def __enter__(self):
        print("Je rentre dans le contexte")
        return "JE suis le return de enter"
    def __exit__(self, exe_type, exe_instance, traceback):
        print(exe_type, exe_instance)
        print("Je sors du contexte")
        return True

with ContextManager() as returnobject:
    print(returnobject)
    raise ValueError("je suis une erreur")
    print("JE suis dans le contexte")
```
