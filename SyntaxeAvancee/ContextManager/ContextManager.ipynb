```python
class ContextManager():
    def __enter__(self):
        print("Je rentre dans le contexte")
        return "JE suis le return de enter"
    def __exit__(self, exe_type, exe_instance, traceback):
        print(exe_type, exe_instance)
        print("Je sors du contexte")
        return True

with ContextManager() as returnobject:
    print(returnobject)
    raise ValueError("je suis une erreur")
    print("JE suis dans le contexte")
```

```python
class FakeDB:
    def begin(self):
        print("Transaction débutée")

    def commit(self):
        print("Transaction validée (commit)")

    def rollback(self):
        print("Transaction annulée (rollback)")
    def close(self):
        print("Closed")

class DBManager():
    def __enter__(self):
        self.db = FakeDB()
        return self.db
    def __exit__(self, exe_type, exe_instance, traceback):
        if exe_type:
            self.db.rollback()
        else:
            self.db.commit()
        self.db.close()
        return True

with DBManager() as db:
    print(db)
    raise ValueError
    
print("end")
```

```python
from contextlib import contextmanager

@contextmanager
def mon_contexte():
    try:
        print("preparation  environnement")
        yield "RESSOURCE"  
  
    finally:
        print("end")


with mon_contexte() as ressource:
    print(f"Utilisation de la {ressource}")
```

```python
import requests
try:
    session = requests.Session()
    headers = {"Accept": "text/plain"}
    response = session.get("https://icanhazdadjoke.com", headers = headers)
    if response.status_code == 200:
        print(response.text)
finally:
    session.close()
```

```python
class RequestsManager:
    def __enter__(self):
         self.session = requests.Session()
         return self.session
    def __exit__(self, exc_type, exc_instance, traceback):
        self.session.close()
        return True
        
with RequestsManager() as session:
    headers = {"Accept": "text/plain"}
    response = session.get("https://icanhazdadjoke.com", headers = headers)
    if response.status_code == 200:
        print(response.text)
```
