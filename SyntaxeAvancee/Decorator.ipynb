```python
import time
def cache(func):
    cache = {}
    def wrapper(*args):
        if args in cache:
            print("cached")
            return cache[args]
        result = func(*args)
        cache[args] = result
        print(cache)
        return result
    return wrapper

@cache
def function(a, b):
    time.sleep(2)
    return a + b

function(3, 1)
function(4, 5)
function(3, 1)
```

```python
import time
def cache_limit(limit):
    def cache(func):
        cache = {}
        def wrapper(*args):
            if args in cache:
                print("cached")
                return cache[args]
            result = func(*args)
            cache[args] = result
            if len(cache) > limit:
                cache.pop(list(cache.keys())[0])
            print(cache)
            return result
        return wrapper
    return cache

@cache_limit(2)
def function(a, b):
    time.sleep(2)
    return a  + b
    
function(3, 1)
function(4, 5)
function(3, 1)
function(-8, -8)
function(-99, -99)
```

```python
def validate_types(types):
    def decorator(func):
        def wrapper(*args):
            if len(args) != len(types):
                raise ValueError(f"len(args) != len(types): {len(args)} != {len(types)}")
            for arg, type_ in zip(args, types):
                if not isinstance(arg, type_):
                    raise TypeError(f"{arg} is not {type_} is {type(arg)}")
            return func(*args)
        return wrapper
    return decorator

@validate_types(types=[int, str])
def process_data(identifiant, name):
    print(f"Processing {identifiant}: {name}")
    

process_data(3, "OK")
#process_data(3)
process_data(3.5, "OK")
```
