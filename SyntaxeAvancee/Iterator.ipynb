```python
class Counteur:
    def __init__(self, depart):
        self.depart = depart
        self.current = depart
        
    def __next__(self):
        if self.current < 0:
            raise StopIteration
        self.current -= 1
        return self.current + 1

    def __iter__(self):
        return self

count = Counteur(11)
for number in count:
    print(number)
```

```python
class CounteurState:
    def __init__(self, depart):
        self.depart = depart
        self.current = depart
        
    def __next__(self):
        if self.current < 0:
            raise StopIteration
        self.current -= 1
        return self.current + 1
        
class Counteur:
    def __init__(self, depart):
        self.depart = depart

    def __iter__(self):
        return CounteurState(self.depart)

count = Counteur(11)
for number in count:
    print(number)
    
for number in count:
    print(number)
```

```python
class PrimeIteratorState:
    def __init__(self, limit):
        self.limit = limit
        self.current = 2
        
    def __next__(self):
        while self.current <= self.limit:
            is_prime = True
            for num in (2, self.current - 1):
                if (self.current % num) == 0:
                    is_prime = False
                    break
            if is_prime:
                prime = self.current
                self.current += 1
                return prime
            self.current += 1
        raise StopIteration
            
        
class PrimeIterator:
    def __init__(self, depart):
        self.depart = depart

    def __iter__(self):
        return PrimeIteratorState(self.depart)

primeiterator = PrimeIterator(20)
for number in primeiterator:
    print(number)

for number in primeiterator:
    print(number)
```
